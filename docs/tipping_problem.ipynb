# Tipping Problem

```python
import altair as alt

from mostly.fuzzy_rules.fuzzy_rule import FuzzyRule
from mostly.fuzzy_rules.logical_operators import And, Is, Or
from mostly.inference.mamdani import MamdaniFIS
from mostly.linguistic_variable import LinguisticVariable
from mostly.membership_functions import MFTriangular
from mostly.plotting.altair.plot_fis_inputs import plot_inference_inputs
from mostly.plotting.altair.plot_fis_outputs import plot_inference_outputs
from mostly.plotting.altair.plot_linguistic_variable import plot_linguistic_variable

alt.renderers.set_embed_options(actions=False)
```

## Define Linguistic Variables

### Service

```python
service = LinguisticVariable(
    concept="Service",
    uod=(0, 10),
    fuzzy_sets={
        "poor": MFTriangular(a=0, b=0, c=5),
        "average": MFTriangular(a=0, b=5, c=10),
        "good": MFTriangular(a=5, b=10, c=10),
    },
)
```

```python
plot_linguistic_variable(service)
```

### Food Quality

```python
quality = LinguisticVariable(
    concept="Quality",
    uod=(0, 10),
    fuzzy_sets={
        "poor": MFTriangular(a=0, b=0, c=5),
        "average": MFTriangular(a=0, b=5, c=10),
        "good": MFTriangular(a=5, b=10, c=10),
    },
)
```

```python
plot_linguistic_variable(quality)
```

### Target Variable "Tip"

```python
tip = LinguisticVariable(
    concept="Tip",
    uod=(0, 25),
    fuzzy_sets={
        "low": MFTriangular(a=0, b=0, c=13),
        "medium": MFTriangular(a=0, b=13, c=25),
        "high": MFTriangular(a=13, b=25, c=25),
    },
)
```

```python
plot_linguistic_variable(tip)
```

## Define Fuzzy Rules

```python
rules = [
    FuzzyRule(
        antecedent=Or(
            [
                Is("service", "poor"),
                Is("quality", "poor"),
            ]
        ),
        consequences={"tip": "low"},
    ),
    FuzzyRule(
        antecedent=Is("service", "average"),
        consequences={"tip": "medium"},
    ),
    FuzzyRule(
        antecedent=Or(
            [
                Is("service", "good"),
                Is("quality", "good"),
            ]
        ),
        consequences={"tip": "high"},
    ),
]
```

## Setup Mamdani Fuzzy Inference System

```python
fis = MamdaniFIS(
    input_variables={"service": service, "quality": quality},
    output_variables={"tip": tip},
    fuzzy_rules=rules,
)
```

### Inference

```python
fis.infer(crisp_inputs={"service": 9.8, "quality": 6.5})
```

### Visual Inference

```python
plot_inference_inputs(fis=fis, crisp_inputs={"service": 5.8, "quality": 6.5}, resolution=1000)
```

```python
plot_inference_outputs(fis=fis, crisp_inputs={"service": 5.8, "quality": 6.5})
```
