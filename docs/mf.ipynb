# Membership Functions

A membership function (*MF*) assigns a membership value between 0 and 1 to each point in the input space, indicating the degree to which that point belongs to a fuzzy set. This input space is called the universe of discourse (*UoD*). A degree of membership of 0 means "not a member" and 1 means "fully a member" of the fuzzy set.

```python
import altair as alt

from mostly.membership_functions import (
    MFCrisp,
    MFPolynomialCombined,
    MFPolynomialS,
    MFPolynomialZ,
    MFSigmoidal,
    MFSigmoidalCombined,
    MFTrapezoidal,
    MFTriangular,
)
from mostly.plotting.altair.plot_fuzzy_set import plot_fuzzy_set

alt.renderers.set_embed_options(actions=False)
```

## Piecewise Linear Functions

### Z-shaped

```python
plot_fuzzy_set(
    mf=MFTrapezoidal(a=2, b=2, c=2, d=8),
    term="good",
    uod=(0, 10),
)
```

Use same logic to model fully linear memberships

```python
plot_fuzzy_set(
    mf=MFTrapezoidal(a=0, b=0, c=0, d=10),
    term="good",
    uod=(0, 10),
)
```

### S-Shaped

```python
plot_fuzzy_set(
    mf=MFTrapezoidal(a=2, b=8, c=8, d=8),
    term="good",
    uod=(0, 10),
)
```

### Triangular

```python
plot_fuzzy_set(
    mf=MFTriangular(a=0, b=5, c=10),
    term="good",
    uod=(0, 10),
)
```

Use `MFTriangular` or `MFTrapezoidal` to model a triangular.

```python
plot_fuzzy_set(
    mf=MFTrapezoidal(a=0, b=5, c=5, d=10),
    term="good",
    uod=(0, 10),
)
```

### Trapezoidal

```python
plot_fuzzy_set(
    mf=MFTrapezoidal(a=0, b=3, c=5, d=10),
    term="good",
    uod=(0, 10),
)
```

## Sigmoidal

### Z-Shaped

```python
plot_fuzzy_set(
    mf=MFSigmoidal(a=-2, c=4),
    term="good",
    uod=(0, 10),
)
```

### S-Shaped

```python
plot_fuzzy_set(
    mf=MFSigmoidal(a=2, c=4),
    term="good",
    uod=(0, 10),
)
```

### Combined Difference

```python
plot_fuzzy_set(
    mf=MFSigmoidalCombined(
        mfs=(MFSigmoidal(a=5, c=2), MFSigmoidal(a=5, c=7)),
        method="diff",
    ),
    term="good",
    uod=(0, 10),
)
```

### Combined Multiply

```python
plot_fuzzy_set(
    mf=MFSigmoidalCombined(
        mfs=(MFSigmoidal(a=2, c=3), MFSigmoidal(a=-5, c=8)),
        method="multiply",
    ),
    term="good",
    uod=(0, 10),
)
```

## Polynomial

### Z-Shaped

```python
plot_fuzzy_set(
    mf=MFPolynomialZ(shoulder=2, foot=8),
    term="good",
    uod=(0, 10),
)
```

### S-shaped

```python
plot_fuzzy_set(
    mf=MFPolynomialS(shoulder=8, foot=2),
    term="good",
    uod=(0, 10),
)
```

### Pi-shaped

```python
plot_fuzzy_set(
    mf=MFPolynomialCombined(
        left=MFPolynomialS(shoulder=4, foot=1),
        right=MFPolynomialZ(shoulder=5, foot=9),
    ),
    term="good",
    uod=(0, 10),
)
```

## Crisp

```python
plot_fuzzy_set(
    mf=MFCrisp(left=5, right=10),
    term="good",
    uod=(0, 10),
)
```

### Manual inclusion of boundaries

```python
plot_fuzzy_set(
    mf=MFCrisp(left=5, right=10, include_left=True, include_right=False),
    term="good",
    uod=(0, 10),
)
```
